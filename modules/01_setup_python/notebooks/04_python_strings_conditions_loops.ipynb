# Strings, Conditions, and Loops

Support workflows often inspect text, clean it up, format it for display, and make simple decisions.

```python
message = "  Our checkout integration is timing out for every payment request.  "
clean_message = message.strip()
normalized = clean_message.lower()

print(clean_message)
print(normalized)
```

Use common string methods to inspect and transform text.

```python
subject = "production api timeout"

print(subject.title())
print(subject.replace("api", "API"))
print(subject.startswith("production"))
print("timeout" in subject)
```

Use f-strings to format readable summaries.

```python
ticket_id = "TCK-1004"
queue = "Engineering Escalation"
priority = "urgent"

summary = f"{ticket_id} | {priority.upper()} | {queue}: {subject.title()}"
print(summary)
```

Use `.split()` when you want to break text into pieces, and `.join()` when you want to combine pieces back into text.

```python
raw_tags = "production,api,timeout"
tags = raw_tags.split(",")
display_tags = " | ".join(tags)

print(tags)
print(display_tags)
```

Conditions let code choose a path.

```python
if "timeout" in normalized or "production" in normalized:
    priority = "urgent"
else:
    priority = "normal"

print(priority)
```

Loops apply the same check to many tickets.

```python
tickets = [
    {"id": "TCK-1001", "message": "My card shows two annual charges."},
    {"id": "TCK-1002", "message": "Sign in loops back to the same page."},
    {"id": "TCK-1004", "message": "Production API requests are timing out."},
]

for ticket in tickets:
    text = ticket["message"].lower()
    if "production" in text or "timeout" in text:
        print(ticket["id"], "urgent")
    else:
        print(ticket["id"], "normal")
```
