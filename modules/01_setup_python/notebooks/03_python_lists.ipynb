# Python Lists

A list keeps multiple values in order. Here the values are ticket dictionaries.

```python
tickets = [
    {
        "id": "TCK-1001",
        "subject": "Charged twice for annual plan",
        "message": "My card shows two annual charges.",
    },
    {
        "id": "TCK-1002",
        "subject": "Cannot sign in after password reset",
        "message": "Sign in loops back to the same page.",
    },
    {
        "id": "TCK-1003",
        "subject": "Where do I download invoices?",
        "message": "I need invoices for the last three months.",
    },
]

print(len(tickets))
```

Use indexes to access items by position. Negative indexes count from the end.

```python
first_ticket = tickets[0]
last_ticket = tickets[-1]

print(first_ticket["subject"])
print(last_ticket["subject"])
```

Use slicing to access a range of items.

```python
first_two_tickets = tickets[:2]
tickets_after_first = tickets[1:]

print(len(first_two_tickets))
print(len(tickets_after_first))
```

Loop through every ticket when you want to apply the same operation to the whole list.

```python
for ticket in tickets:
    print(f"{ticket['id']}: {ticket['subject']}")
```

Build a list of values from a list of dictionaries with a list comprehension.

```python
subjects = [ticket["subject"] for ticket in tickets]

print(subjects)
```

Sort a list when order matters. `sorted()` returns a new sorted list.

```python
sorted_subjects = sorted(subjects)

print(sorted_subjects)
print(subjects)
```

A string can become a list with `.split()`. This is useful when a field contains comma-separated labels.

```python
raw_tags = "billing,duplicate-charge,annual-plan"
tags = raw_tags.split(",")

print(tags)
print("billing" in tags)
```

Add items with `.append()` when new data arrives.

```python
tickets.append(
    {
        "id": "TCK-1004",
        "subject": "Production API timeout",
        "message": "Checkout requests are timing out.",
    }
)

print(len(tickets))
print(tickets[-1]["subject"])
```
