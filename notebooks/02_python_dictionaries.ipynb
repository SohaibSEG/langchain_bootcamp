# Python Dictionaries

A dictionary groups related values under names called keys.

```python
ticket = {
    "id": "TCK-1002",
    "customer": "Noah",
    "subject": "Cannot sign in after password reset",
    "message": "The reset link worked, but sign in loops back to the same page.",
    "channel": "chat",
}

print(ticket)
```

Required access uses square brackets. Use it when the field must exist.

```python
print(ticket["subject"])
print(ticket["message"])
```

Safe access uses `.get()`. Use it when the field might be missing.

```python
priority = ticket.get("priority", "normal")
assigned_agent = ticket.get("assigned_agent")

print(priority)
print(assigned_agent)
```

You can unpack selected fields into variables when you want shorter names.

```python
ticket_id = ticket["id"]
subject = ticket["subject"]
channel = ticket["channel"]

print(ticket_id, channel, subject)
```

Dictionary destructuring/unpacking with `**` creates a new dictionary from existing fields plus changes.

```python
triaged_ticket = {
    **ticket,
    "priority": "high",
    "queue": "Account",
}

print(triaged_ticket)
print(ticket)
```

Loop through keys and values with `.items()` when you need to inspect the whole dictionary.

```python
for key, value in triaged_ticket.items():
    print(key, "=>", value)
```

JSON is a common text format for storing and sending dictionaries. Use `json.dumps()` to convert a dictionary to a JSON string, and `json.loads()` to convert JSON text back to a dictionary.

```python
import json

ticket_json = json.dumps(triaged_ticket, indent=2)
print(ticket_json)

ticket_from_json = json.loads(ticket_json)
print(ticket_from_json["queue"])
```

Add or update fields as the workflow learns more.

```python
ticket["status"] = "new"
ticket["priority"] = "normal"

print(ticket)
```
