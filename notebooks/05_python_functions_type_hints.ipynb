# Functions and Type Hints

Functions let us name reusable logic. Type hints describe the expected shape of values.

```python
def summarize_ticket(ticket: dict[str, str]) -> str:
    return f"{ticket['id']}: {ticket['subject']}"


ticket = {
    "id": "TCK-1002",
    "subject": "Cannot sign in after password reset",
    "message": "Sign in loops back to the same page.",
}

print(summarize_ticket(ticket))
```

A function can return a decision.

```python
def priority_for(ticket: dict[str, str]) -> str:
    text = f"{ticket['subject']} {ticket['message']}".lower()
    if "production" in text or "timeout" in text or "charged twice" in text:
        return "urgent"
    return "normal"


print(priority_for(ticket))
```

Function arguments can be passed by position or by name.

```python
def format_ticket(ticket_id: str, subject: str, priority: str = "normal") -> str:
    return f"{ticket_id} | {priority.upper()} | {subject}"


positional_result = format_ticket("TCK-1004", "Production API timeout", "urgent")
named_result = format_ticket(
    ticket_id="TCK-1004",
    subject="Production API timeout",
    priority="urgent",
)
default_result = format_ticket("TCK-1003", "Where do I download invoices?")

print(positional_result)
print(named_result)
print(default_result)
```

Named arguments are clearer when a function has several inputs or optional values.

```python
reply_summary = format_ticket(
    subject="Charged twice for annual plan",
    ticket_id="TCK-1001",
    priority="high",
)

print(reply_summary)
```

Apply the same functions to many tickets.

```python
tickets = [
    ticket,
    {
        "id": "TCK-1004",
        "subject": "Production API timeout",
        "message": "Checkout requests are timing out.",
    },
]

for item in tickets:
    print(summarize_ticket(item), priority_for(item))
```
