# Classes, Inheritance, Overrides, and Abstract Classes

Classes group data and behavior. Use them when a concept has a stable shape in your program.

```python
class SupportTicket:
    def __init__(self, ticket_id: str, subject: str, message: str) -> None:
        self.ticket_id = ticket_id
        self.subject = subject
        self.message = message

    def summary(self) -> str:
        return f"{self.ticket_id}: {self.subject}"

    def queue(self) -> str:
        return "Support"


ticket = SupportTicket(
    "TCK-1003",
    "Where do I download invoices?",
    "I need invoices for the last three months.",
)

print(ticket.summary())
print(ticket.queue())
```

Inheritance lets a specialized class reuse behavior from a general class.

```python
class EscalatedTicket(SupportTicket):
    def __init__(self, ticket_id: str, subject: str, message: str, escalation_queue: str) -> None:
        super().__init__(ticket_id, subject, message)
        self.escalation_queue = escalation_queue

    def escalation_summary(self) -> str:
        return f"{self.summary()} -> {self.escalation_queue}"


escalated = EscalatedTicket(
    "TCK-1004",
    "Production API timeout",
    "Checkout requests are timing out.",
    "Engineering Escalation",
)

print(escalated.summary())
print(escalated.escalation_summary())
```

A subclass can override a method when it needs different behavior with the same method name.

```python
class BillingTicket(SupportTicket):
    def queue(self) -> str:
        return "Billing"


class EngineeringTicket(SupportTicket):
    def queue(self) -> str:
        return "Engineering Escalation"

    def summary(self) -> str:
        return f"{self.ticket_id}: URGENT - {self.subject}"


billing_ticket = BillingTicket(
    "TCK-1001",
    "Charged twice for annual plan",
    "My card shows two annual charges.",
)
engineering_ticket = EngineeringTicket(
    "TCK-1004",
    "Production API timeout",
    "Checkout requests are timing out.",
)

print(billing_ticket.summary())
print(billing_ticket.queue())
print(engineering_ticket.summary())
print(engineering_ticket.queue())
```

Abstract classes define behavior that subclasses must provide. They are useful when several classes should share the same interface, but each class needs its own implementation.

```python
from abc import ABC, abstractmethod


class ReplyStrategy(ABC):
    @abstractmethod
    def draft_reply(self, ticket: SupportTicket) -> str:
        pass


class StandardReplyStrategy(ReplyStrategy):
    def draft_reply(self, ticket: SupportTicket) -> str:
        return f"Thanks for contacting us about '{ticket.subject}'. We will review it shortly."


class EscalationReplyStrategy(ReplyStrategy):
    def draft_reply(self, ticket: SupportTicket) -> str:
        return f"We received your urgent issue '{ticket.subject}' and are escalating it now."


standard_strategy = StandardReplyStrategy()
escalation_strategy = EscalationReplyStrategy()

print(standard_strategy.draft_reply(billing_ticket))
print(escalation_strategy.draft_reply(engineering_ticket))
```

You cannot create an abstract class directly if it still has abstract methods.

```python
try:
    broken_strategy = ReplyStrategy()
except TypeError as error:
    print(error)
```
